```python
import matplotlib.pyplot as plt
from astropy.io import fits
import numpy as np
import glob
```

```python
file_names = [f"M13_0000{i}.fits" for i in range(1, 6)]
image_data_list = []

for file in file_names:
    try:
        with fits.open(file) as hdul:
            data = hdul[0].data
            image_data_list.append(data)
            print(f"Loaded {file}")
    except FileNotFoundError:
        print(f"Could not find {file}, please check path.")
```

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i, ax in enumerate(axes):
    if i < len(image_data_list):
        im = ax.imshow(image_data_list[i], origin="lower")
        ax.set_title(f"Image {i+1}")
plt.show()
```

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for i, ax in enumerate(axes):
    if i < len(image_data_list):
        im = ax.imshow(image_data_list[i], origin="lower", vmin=0, vmax=2000)
        ax.set_title(f"Image {i+1}")
        fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.show()
```

```python
# Plot Histograms
for i, data in enumerate(image_data_list):
    plt.figure()
    plt.hist(data.ravel(), bins=50)
    plt.title(f"Histogram for Image {i+1}")
    plt.xlabel("Pixel Value")
    plt.ylabel("Frequency")
    plt.yscale('log') 
    plt.show()
    
    print(f"Image {i+1} Mean: {np.mean(data)}")
    print(f"Image {i+1} Median: {np.median(data)}")
```

```python
# 3.4

stacked_data = np.array(image_data_list)
combined_image = np.median(stacked_data, axis=0)

print(f"Combined Median Value: {np.median(combined_image)}")
```

```python
plt.figure()
plt.hist(combined_image.ravel(), bins=50)
plt.title("Combined Image Histogram")
plt.show()
```

```python
vmin_val = 900  
vmax_val = 1100 

plt.figure(figsize=(6, 6))
plt.imshow(combined_image, origin="lower", vmin=vmin_val, vmax=vmax_val, cmap='gray')
plt.colorbar()
plt.title("Combined Image (Normalized)")
plt.show()
```

```python
fits.writeto("combined.fits", combined_image, overwrite=True)
print("Saved combined.fits")
```
