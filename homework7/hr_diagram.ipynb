```python
# 3 - HR Diagram
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
print(f"Data shape: {data.shape}")
print(f"There are {data.shape[0]} rows in the dataset.")
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
```

```python
min_mag = np.min(magnitude)
max_mag = np.max(magnitude)
min_mag_star_index = np.argmin(magnitude)
max_mag_star_index = np.argmax(magnitude)

print(f"Magnitude Min: {min_mag} (Star: {min_mag_star_index})")
print(f"Magnitude Max: {max_mag} (Star: {max_mag_star_index})")
```

```python
min_temp = np.min(temperature)
max_temp = np.max(temperature)
min_temp_star_index = np.argmin(temperature)
max_temp_star_index = np.argmax(temperature)
    
print(f"Temperature Min: {min_temp} K (Star: {min_temp_star_index})")
print(f"Temperature Max: {max_temp} K (Star: {max_temp_star_index})")
```

```python
print(f"\nThe brightest star is at index {min_mag_star_index} with the smallest magnitude: {min_mag}.")
print(f"The dimmest star is at index {max_mag_star_index} with the largest magnitude: {max_mag}.")
```

```python
# 3.3 - Plot
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8, 6))
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

# Brightest star
brightest_log_temp = log_temperature[min_mag_star_index]
brightest_log_mag = log_magnitude[min_mag_star_index]
    
# Dimmest star
dimmest_log_temp = log_temperature[max_mag_star_index]
dimmest_log_mag = log_magnitude[max_mag_star_index]
    
# Hottest star
hottest_log_temp = log_temperature[max_temp_star_index]
hottest_log_mag = log_magnitude[max_temp_star_index]

# Coldest star
coldest_log_temp = log_temperature[min_temp_star_index]
coldest_log_mag = log_magnitude[min_temp_star_index]


plt.scatter(brightest_log_temp, brightest_log_mag, 
                color='black', marker='*', s=200, label='Brightest Star', zorder=5)
plt.scatter(dimmest_log_temp, dimmest_log_mag, 
                color='purple', marker='*', s=200, label='Dimmest Star', zorder=5)
plt.scatter(hottest_log_temp, hottest_log_mag, 
                color='green', marker='*', s=200, label='Hottest Star', zorder=5)
plt.scatter(coldest_log_temp, coldest_log_mag, 
                color='orange', marker='*', s=200, label='Coldest Star', zorder=5)
plt.scatter(
        log_temperature, 
        log_magnitude, 
        alpha=0.5, 
        label='All Stars'
)

plt.legend(fontsize=10)
plt.title("Hertzsprung-Russell Diagram", fontsize=18)
plt.xlabel("Temperature (log K)", fontsize=14)
plt.ylabel("Magnitude", fontsize=14)
plt.show()
```
