```python
# 2 - Matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    """
    Returns y = A * sin(w * x)
    
    Args:
        x (np.array): Input values
        A (float): Amplitude
        w (float): Angular frequency
    """
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
```

```python
amplitudes = [0.5, 1, 1.5, 2, 2.5]
frequencies = [1, 2, 3, 4, 5]
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(25, 5))
fig.suptitle('Sine Waves with Varying Parameters', fontsize=20)

# Plot 1
ax1.set_title("Varying Amplitude", fontsize=16)
ax1.set_xlabel("x", fontsize=14)
ax1.set_ylabel("y = A * sin(x)", fontsize=14)
for A in amplitudes:
    y = make_sine_wave(x, A, w=1)
    ax1.plot(x, y, label=f'A={A}')
ax1.legend(loc='upper right', fontsize=12)

# Plot 2
ax2.set_title("Varying Frequency", fontsize=16)
ax2.set_xlabel("x", fontsize=14)
ax2.set_ylabel("y = sin(w * x)", fontsize=14)
for w in frequencies:
    y = make_sine_wave(x, A=1, w=w)
    ax2.plot(x, y, label=f'w={w}')
ax2.legend(loc='lower left', fontsize=12)

# Plot 3
ax3.set_title("Varying Amplitude and Frequency", fontsize=16)
ax3.set_xlabel("x", fontsize=14)
ax3.set_ylabel("y = A * sin(w * x)", fontsize=14)
for A, w in zip(amplitudes, frequencies):
    y = make_sine_wave(x, A, w)
    ax3.plot(x, y, label=f'A={A}, w={w}')
ax3.legend(loc='lower left', fontsize=12)

plt.show()
```
