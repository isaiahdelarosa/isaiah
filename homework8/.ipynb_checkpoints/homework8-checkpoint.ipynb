```python
import pandas as pd
```

```python
state_data = pd.read_csv("../../course_assignments/homework8/state_data.csv")
```

```python
print("Shape:", state_data.shape)
print("Number of rows:", state_data.shape[0])
print("Number of columns:", state_data.shape[1])
```

```python
display(state_data.head())
display(state_data.tail())
print("\nFirst state:", state_data.iloc[0]['state'])
print("Last state:", state_data.iloc[-1]['state'])
print("Total entries:", len(state_data))
print("There are more than 50 because it includes territories.")
```

```python
median_lat = state_data['latitude'].median()
median_lon = state_data['longitude'].median()
print(f"Median Latitude: {median_lat:.2f}, Median Longitude: {median_lon:.2f}")
```

```python
state_data['lat_diff'] = abs(state_data['latitude'] - median_lat)
```

```python
state_data['lon_diff'] = abs(state_data['longitude'] - median_lon)
```

```python
print("New shape:", state_data.shape)
```

```python
closest_lat = state_data.loc[state_data['lat_diff'].idxmin()]
farthest_lat = state_data.loc[state_data['lat_diff'].idxmax()]
closest_lon = state_data.loc[state_data['lon_diff'].idxmin()]
farthest_lon = state_data.loc[state_data['lon_diff'].idxmax()]

print(f"Closest to median latitude: {closest_lat['state']} ({closest_lat['lat_diff']:.2f})")
print(f"Farthest from median latitude: {farthest_lat['state']} ({farthest_lat['lat_diff']:.2f})")
print(f"Closest to median longitude: {closest_lon['state']} ({closest_lon['lon_diff']:.2f})")
print(f"Farthest from median longitude: {farthest_lon['state']} ({farthest_lon['lon_diff']:.2f})")
```

```python
state_data_copy = state_data.copy()

print("Original DataFrame:")
display(state_data.head())
print("Copied DataFrame:")
display(state_data_copy.head())

# I found the .copy() function on the website under the "Essential Basic Functionality" tab.
# It duplicates both the data and the structure so changes made won't affect the original data frame.
# This is useful because it could prevent unintentional changes to the original data.
```
